```python
from langchain_community.document_loaders import (
    PyPDFLoader,
    Docx2txtLoader
)

import os

files = [
    "RAG_docs/smoking_and_diabetes.pdf",
    "RAG_docs/On-your-way-to-preventing-type-2-diabetes.pdf",
    "RAG_docs/WHO-HEP-HPR-RUN-2021.2-eng.pdf",
    "RAG_docs/Business_Recommendations_RAG 1.docx",
    "RAG_docs/Detailed_Population_Health_Strategy_Guide.docx",
    "RAG_docs/Detailed_Population_Segmentation_Knowledge_Base.docx",
    "RAG_docs/Detailed_Preventive_Intervention_Playbooks.docx",
    "RAG_docs/Doc03_High_Risk_Member_Guide 2.docx",
    "RAG_docs/Wellness_Campaign_Recommendation_Catalog 2 2.docx"
]

docs = []

for file in files:

    if file.endswith(".pdf"):
        loader = PyPDFLoader(file)

    elif file.endswith(".docx"):
        loader = Docx2txtLoader(file)

    loaded_docs = loader.load()

    # Add metadata to every document
    for doc in loaded_docs:
        doc.metadata["source_file"] = os.path.basename(file)
        doc.metadata["document_type"] = (
            "pdf" if file.endswith(".pdf") else "docx"
        )

    docs.extend(loaded_docs)

print(f"Documents Loaded: {len(docs)}")

# Verify metadata
print(docs[0].metadata)
```

chunk documents

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = splitter.split_documents(docs)

print(f"Total Chunks: {len(chunks)}")

# Check metadata is still present
print(chunks[0].metadata)
```

create embeddings

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5" #all-MiniLM-L6-v2
)
```

create FAISS database

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    chunks,
    embedding_model
)
```

save the data

```python
vectorstore.save_local(
    "population_health_index"
)
```

Build retriever

```python
retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k":7,
        "fetch_k":30
    }
)
#top 5 relevant chunks will be retrieved for each query
```

```python
def generate_member_recommendation(member):
    
    # ---------------------------------------
    # Build Retrieval Query
    # ---------------------------------------
    
    query = f"""
    Cluster: {member['Cluster']}
    
    Age: {member['Age']}
    
    BMI: {member['BMI']}
    
    Blood Glucose: {member['Blood_Glucose']}
    
    Smoking Status: {member['Smoking_Status']}
    
    Hypertension: {member['Hypertension']}
    
    Risk Category: {member['Risk_Category']}
    
    Need:
    
    - diabetes prevention recommendations
    - smoking cessation recommendations
    - weight management interventions
    - wellness program recommendations
    - intervention playbooks
    - preventive screenings
    - care management actions
    - business recommendations
    """

    # ---------------------------------------
    # Retrieve Documents
    # ---------------------------------------

    results = retriever.invoke(query)

    # ---------------------------------------
    # Remove Low-Risk Noise
    # ---------------------------------------

    filtered_docs = []

    for doc in results:

        if "Healthy Low-Risk Members" in doc.page_content:
            continue

        filtered_docs.append(doc)

    # ---------------------------------------
    # Build Context
    # ---------------------------------------

    context = "\n\n".join(
        [doc.page_content for doc in filtered_docs]
    )

    # ---------------------------------------
    # Build Source List
    # ---------------------------------------

    sources = "\n".join(
        list(
            set(
                doc.metadata.get(
                    "source_file",
                    "Unknown"
                )
                for doc in filtered_docs
            )
        )
    )

    # ---------------------------------------
    # Fill Prompt
    # ---------------------------------------

    filled_prompt = template.format(
        age=member['Age'],
        bmi=member['BMI'],
        glucose=member['Blood_Glucose'],
        smoking=member['Smoking_Status'],
        hypertension=member['Hypertension'],
        risk=member['Risk_Category'],
        probability=member['Predicted_Probability'],
        context=context,
        sources=sources
    )

    # ---------------------------------------
    # Generate Recommendation
    # ---------------------------------------

    response = llm.invoke(
        filled_prompt
    )

    return response.content
```

create recommendation function

```python
#rename some cols and clean up the dataframe
# Clean columns

rag_df = rag_df.rename(
    columns={
        "Risk_Category_x": "Risk_Category"
    }
)

# Remove members without cluster information

rag_df = rag_df.dropna(
    subset=["Cluster_Name"]
)

print(
    rag_df[
        [
            "Member_ID",
            "Cluster_Name",
            "Risk_Category",
            "Risk_Probability"
        ]
    ].head()
)
```

```python
def generate_member_recommendation(member):

    query = f"""
    Cluster: {member['Cluster_Name']}

    Age: {member['Age']}

    BMI: {member['BMI']}

    Blood Glucose: {member['Glucose']}

    Smoking Status: {member['Smoking_Status']}

    Hypertension:
    {member['Existing_Hypertension_Flag']}

    Risk Category:
    {member['Risk_Category']}

    Risk Probability:
    {member['Risk_Probability']}

    Need:

    - clinical recommendations

    - wellness program recommendations

    - intervention playbooks

    - preventive screenings

    - care management actions

    - business recommendations

    - population health interventions
    """

    # retrieve docs
    results = retriever.invoke(query)

    # remove noisy chunk
    filtered_docs = []

    for doc in results:

        if "Healthy Low-Risk Members" in doc.page_content:
            continue

        filtered_docs.append(doc)

    # build context
    context = "\n\n".join(
        [doc.page_content for doc in filtered_docs]
    )

    # source list
    sources = "\n".join(
        list(
            set(
                doc.metadata.get(
                    "source_file",
                    "Unknown"
                )
                for doc in filtered_docs
            )
        )
    )

    # fill prompt
    filled_prompt = template.format(
        age=member["Age"],
        bmi=member["BMI"],
        glucose=member["Glucose"],
        smoking=member["Smoking_Status"],
        hypertension=member[
            "Existing_Hypertension_Flag"
        ],
        risk=member["Risk_Category"],
        probability=member[
            "Risk_Probability"
        ],
        context=context,
        sources=sources
    )

    response = llm.invoke(
        filled_prompt
    )

    return response.content
```

store recommendations for 50 members

```python
high_risk_members = rag_df[
    rag_df["Risk_Probability"] > 0.80
]

recommendations = []

for _, member in high_risk_members.head(25).iterrows():

    try:

        rec = generate_member_recommendation(
            member
        )

        recommendations.append({

            "Member_ID":
                member["Member_ID"],

            "Risk_Probability":
                member["Risk_Probability"],

            "Risk_Category":
                member["Risk_Category"],

            "Cluster_Name":
                member["Cluster_Name"],

            "AI_Recommendation":
                rec
        })

        print(
            f"Completed "
            f"{member['Member_ID']}"
        )

    except Exception as e:

        print(e)
```

add the records to the csv file

```python
import pandas as pd

recommendation_df = pd.DataFrame(
    recommendations
)

recommendation_df.to_csv(
    "member_recommendations.csv",
    index=False
)
```
